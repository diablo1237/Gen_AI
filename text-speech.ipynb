```python
!pip install gTTS
from gtts import gTTS
from IPython.display import Audio
# Text to convert
text = """
I am Text to speech converter. I convert text to speech
"""

# Language
language = 'en'

# Create TTS object
tts = gTTS(text=text, lang=language, slow=False)

# Save audio file
tts.save("speech.mp3")

print("Speech saved as speech.mp3")
Audio("speech.mp3", autoplay=True)
```

```python
# Install required libraries
!pip install -q diffusers transformers accelerate torch sentencepiece
!pip install "click>7.1"

# Import libraries
import torch
from diffusers import StableDiffusionPipeline
from IPython.display import display

# Load the pre-trained Stable Diffusion model
model_id = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
)

# Use GPU if available
pipe = pipe.to("cuda")

# Enter your text prompt
prompt = input("Enter your prompt: ")

# Generate the image
image = pipe(prompt).images[0]

# Display the image
display(image)

# Save the image
image.save("generated_image.png")

print("Image saved as generated_image.png")
```
